```python
customers["age_valid"] = customers["age"] > 0

print("Invalid Age Records:")
print(customers[customers["age_valid"] == False])
```

```python
today = pd.Timestamp.today()

customers["age"] = (
    today.year - customers["date_of_birth"].dt.year
    - (
        (customers["date_of_birth"].dt.month > today.month)
        |
        (
            (customers["date_of_birth"].dt.month == today.month)
            & (customers["date_of_birth"].dt.day > today.day)
        )
    ).astype(int)
)
```
